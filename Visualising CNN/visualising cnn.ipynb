```python
# 1. Target the actual first convolutional layer in your VGG model
conv_layer = model.get_layer('block1_conv1') 
conv_filters, biases = conv_layer.get_weights()

# This will print a 4D shape like (3, 3, 3, 64)
print("Filters shape:", conv_filters.shape) 

# 2. Run your visual subplots loop
import matplotlib.pyplot as pyplot
n_filters = 6
ix = 1
fig = pyplot.figure(figsize=(15, 10))

for i in range(n_filters):
    # Slice the 4D filter matrix cleanly
    f = conv_filters[:, :, :, i] 
    
    # Loop through the 3 color channels (Red, Green, Blue input channels)
    for j in range(3):
        pyplot.subplot(n_filters, 3, ix)
        pyplot.imshow(f[:, :, j], cmap='gray')
        pyplot.xticks([])
        pyplot.yticks([])
        ix += 1

pyplot.show()
```
